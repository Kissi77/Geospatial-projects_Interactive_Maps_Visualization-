```python
# Install required packages (run this in your terminal, not inside the Python file)
# pip install geopandas plotly pydeck

# -----------------------------
# PLOTLY INTERACTIVE WORLD MAP
# -----------------------------
import plotly.express as px

# Create a simple choropleth map using ISO-3 country codes
fig = px.choropleth(
    locations=("DEU", "USA", "GHA", "CHN"),   # List of countries to highlight
    locationmode="ISO-3"                      # Tells plotly to interpret codes as ISO-3
)

# Add world country boundaries
fig.update_layout(
    geo=dict(showcountries=True)
)

# Display the interactive Plotly map
fig.show()


# -----------------------------
# PYDECK INTERACTIVE GLOBE VIEW
# -----------------------------
import pydeck as pdk

# Set the initial view of the globe (centered at 0° latitude/longitude)
view = pdk.ViewState(
    latitude=0,
    longitude=0,
    zoom=0.5
)

# Create an empty Scatterplot layer (can be replaced with real data later)
layer = pdk.Layer(
    "ScatterplotLayer",
    data=[]   # No data yet
)

# Build and display the deck.gl visualization
deck = pdk.Deck(
    layers=[layer],
    initial_view_state=view
)

deck.show()
```
