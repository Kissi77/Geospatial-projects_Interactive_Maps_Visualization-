```python
import geopandas as gpd
from geopy.distance import geodesic
import folium

# -----------------------------
# 1. CREATE A SIMPLE CITY DATASET
# -----------------------------
# Define two cities with their latitude and longitude values
data = {
    'City': ['Dusseldorf', 'Bonn'],
    'Latitude': [51.2230, 50.7374],
    'Longitude': [6.7825, 7.0982]
}

# Convert the data into a GeoDataFrame
# This creates geometry points from longitude and latitude
gdf = gpd.GeoDataFrame(
    data,
    geometry=gpd.points_from_xy(data["Longitude"], data["Latitude"])
)


# -----------------------------
# 2. CALCULATE GEODESIC DISTANCE
# -----------------------------
# Extract coordinate pairs for each city
coords_city1 = (gdf.loc[0, "Latitude"], gdf.loc[0, "Longitude"])
coords_city2 = (gdf.loc[1, "Latitude"], gdf.loc[1, "Longitude"])

# Calculate the geodesic distance in kilometers
distance = geodesic(coords_city1, coords_city2).kilometers

print(f"Distance between {gdf.loc[0,'City']} and {gdf.loc[1,'City']}: {distance:.2f} km")



# -----------------------------
# 3. CREATE AN INTERACTIVE FOLIUM MAP
# -----------------------------
# Find midpoint between the cities to center the map
midpoint = [
    gdf["Latitude"].mean(),
    gdf["Longitude"].mean()
]

# Initialize the map
m = folium.Map(location=midpoint, zoom_start=8)


# -----------------------------
# 4. ADD MARKERS FOR EACH CITY
# -----------------------------
for idx, row in gdf.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=row["City"]
    ).add_to(m)


# -----------------------------
# 5. DRAW A LINE CONNECTING THE CITIES
# -----------------------------
line = folium.PolyLine(
    locations=[
        (gdf.loc[0, "Latitude"], gdf.loc[0, "Longitude"]),
        (gdf.loc[1, "Latitude"], gdf.loc[1, "Longitude"])
    ],
    color="green",
    weight=5.5,
    opacity=1
)

# Add the line to the map
m.add_child(line)

# Display the map in Jupyter Notebook or VS Code
m
```
