```python
from ipywidgets import Label
from ipyleaflet import Map, Marker

# -----------------------------------------
# CREATE AN INTERACTIVE LEAFLET MAP IN JUPYTER
# -----------------------------------------

# Initialize a world map centered at latitude 0°, longitude 0°
# Zoom level 2 gives a global view
m = Map(center=(0, 0), zoom=2)

# Create a label widget that will display the coordinates
label = Label()
display(label)


# -----------------------------------------
# FUNCTION TO HANDLE MAP INTERACTIONS
# -----------------------------------------
def handle_interaction(**kwargs):
    """
    This function listens for click events on the map.
    When the user clicks on the map:
      - It extracts the latitude & longitude of the click
      - Updates the label text with the clicked coordinates
      - Places a marker at the clicked location
    """
    
    # Only react when the user performs a "click" action
    if kwargs.get("type") == "click":
        
        # Extract latitude and longitude from event data
        lat, lon = kwargs.get("coordinates")
        
        # Display clicked coordinates in the label widget
        label.value = f"Clicked on: {lat}, {lon}"
        
        # Create a marker at the clicked location
        marker = Marker(location=(lat, lon))
        
        # Add the marker to the map
        m.add_layer(marker)


# -----------------------------------------
# ATTACH THE CLICK HANDLER TO THE MAP
# -----------------------------------------
m.on_interaction(handle_interaction)

# Display the interactive map
m

```
