```python
from geopy.geocoders import Nominatim

# -----------------------------
# 1. INITIALIZE THE GEOCODER
# -----------------------------
# Nominatim is a geocoding service provided by OpenStreetMap.
# We must provide a unique user_agent string to identify our application.
geolocator = Nominatim(user_agent="geo_coder")

# -----------------------------
# 2. DEFINE THE ADDRESS TO LOOK UP
# -----------------------------
address = "53115 Bonn, Germany"

# -----------------------------
# 3. GEOCODE THE ADDRESS
# -----------------------------
try:
    # Attempt to get location information for the address
    location = geolocator.geocode(address)
    
    if location:
        # If geocoding is successful, print address and coordinates
        print(f"Address: {address}")
        print(f"Latitude: {location.latitude}, Longitude: {location.longitude}")
    else:
        # If no result is found
        print("Location not found.")
        
except Exception as e:
    # Catch any errors (network issues, rate limits, etc.)
    print(f"Error: {e}")
```

```python
from geopy.geocoders import Nominatim

# -----------------------------
# 1. INITIALIZE THE GEOCODER
# -----------------------------
# Nominatim is a geocoding service provided by OpenStreetMap.
# The user_agent parameter is required to identify your application.
geolocator = Nominatim(user_agent="geo_coder")

# -----------------------------
# 2. DEFINE THE COORDINATES
# -----------------------------
# These are the latitude and longitude for which we want the address
latitude = 50.7231139
longitude = 7.0893507

# -----------------------------
# 3. REVERSE GEOCODE THE COORDINATES
# -----------------------------
try:
    # Perform reverse geocoding (coordinates -> human-readable address)
    # exactly_one=True ensures only one result is returned
    location = geolocator.reverse((latitude, longitude), exactly_one=True)
    
    if location:
        # If reverse geocoding is successful, print coordinates and address
        print(f"Latitude: {latitude}, Longitude: {longitude}")
        print(f"Address: {location.address}")
    else:
        # If no address is found for the coordinates
        print(f"No address found for coordinates: Latitude {latitude}, Longitude {longitude}")
        
except Exception as e:
    # Catch any errors (network issues, rate limits, etc.)
    print(f"Error: {e}")

```
